```python
import pandas as pd
from pyspark.sql import SparkSession
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/customers.csv")

spark= SparkSession.builder.getOrCreate()
spark_df=spark.createDataFrame(df)
spark_df.write.format("csv").mode("overwrite").save("/Volumes/ecommerce_analytics/bronze/raw_data/customer")
```

```python
spark= SparkSession.builder.getOrCreate()

tables=  ["customers", "sales","sales_orders"]

base_url = "https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/"

base_path = "/Volumes/ecommerce_analytics/bronze/raw_data/"

for table in tables:
    print(f"procesing the table {table}")
    file_url = f"{base_url}{table}.csv"
    print("processing the url",file_url)
    
    df=pd.read_csv(file_url)
    spark_df=spark.createDataFrame(df)
    out_put=f"{base_path}{table}/"
    spark_df.write.format("csv").mode("overwrite").option("header", True).save(out_put)
    print(f"writing{table}to{out_put}")
    print("Al table proccessed")
 

```
